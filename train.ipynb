```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
import torchvision
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from torchvision import transforms, models
```

```python
#이미지 가공

Data_set = '/content/drive/MyDrive/AICOSS 2026 WE-Meet/data/processed'

training = transforms.Compose([transforms.Resize((224, 224)),
                                transforms.ToTensor(),
                                transforms.Normalize([0.485, 0.456, 0.406],[0.229,0.224,0.225])])

dataset = torchvision.datasets.ImageFolder(root=Data_set, transform=training)
dataloaders = DataLoader(dataset, batch_size = 20, shuffle=True)
print(dataset.classes)
```

```python
#모델 준비
model = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
```

```python
#이미지 시각화
def imshow(inp, title=None): #inp = input, 텐서 이미지 받음, title = 선택...
    inp = inp.numpy().transpose((1,2,0))
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    inp = std*inp+mean # 정규화 되돌리기!
    inp = np.clip(inp,0,1) # 색상 표현시 값이 0보다 작거나 1보다 크면 깨질 수 있음, 0보다 작은 건 0으로, 1보다 큰건 1로
    plt.imshow(inp)
    if title is not None:
        plt.title(title)
    plt.pause(0.001) # 실시간으로 그림 띄우고 쉬었다가 다음코드로 넘어가기
```

```python
#손실함수와 옵티마이저 세팅
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr =0.001) #모델 가중치 수정 결정
```

```python
#훈련
device = torch.accelerator.current_accelerator().type if torch.accelerator.is_available() else "cpu"

def train_model(model, criterion, optimizer, num_epochs=5):
    for epoch in range(num_epochs):
        running_loss = 0.0
        print(f'{epoch}/{num_epochs}')
   #여기부터는 모르겠음....  너무어려워용 ㅠㅠ
    for phase in training:
        model.train()
    for inputs, labels in dataloaders[phase]:
        inputs = inputs.to(device)
        labels = labels.to(device)

train_model(model,criterion,optimizer)
```
